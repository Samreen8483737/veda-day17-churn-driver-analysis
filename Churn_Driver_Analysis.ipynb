```python
import pandas as pd
import numpy as np

# 1. Create a sample 'Telco Churn' dataset
np.random.seed(42)
n = 1000
data = {
    'CustomerID': range(1, n+1),
    'Contract': np.random.choice(['Month-to-month', 'One year', 'Two year'], p=[0.5, 0.25, 0.25], size=n),
    'TechSupport': np.random.choice(['Yes', 'No', 'No internet'], p=[0.3, 0.5, 0.2], size=n),
    'Tenure_Months': np.random.randint(1, 72, size=n)
}
df = pd.DataFrame(data)

# (Simulating churn associations for the exercise)
def assign_churn(row):
    prob = 0.1
    if row['Contract'] == 'Month-to-month': prob += 0.3
    if row['TechSupport'] == 'No': prob += 0.2
    if row['Tenure_Months'] < 12: prob += 0.2
    return np.random.choice(['Yes', 'No'], p=[min(prob, 0.9), max(1-prob, 0.1)])

df['Churn'] = df.apply(assign_churn, axis=1)

# Convert Churn to binary (1 for Yes, 0 for No) for easy math
df['Churn_Num'] = df['Churn'].apply(lambda x: 1 if x == 'Yes' else 0)

# 2. Segment Analysis: Churn Rate by Contract Type
contract_churn = (df.groupby('Contract')['Churn_Num'].mean() * 100).round(2)
print("--- Churn Rate by Contract Type ---")
print(contract_churn.astype(str) + '%')

# 3. Segment Analysis: Churn Rate by Tech Support
tech_churn = (df.groupby('TechSupport')['Churn_Num'].mean() * 100).round(2)
print("\n--- Churn Rate by Tech Support ---")
print(tech_churn.astype(str) + '%')

# 4. Driver Ranking (Identifying the strongest association)
contract_spread = contract_churn.max() - contract_churn.min()
tech_spread = tech_churn.max() - tech_churn.min()

print("\n--- Driver Ranking (Max Difference in Churn Rate) ---")
print(f"1. Contract Type Impact: {contract_spread:.2f}% spread")
print(f"2. Tech Support Impact: {tech_spread:.2f}% spread")
```
